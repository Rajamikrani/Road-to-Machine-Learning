# Basic Linear Regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
study_hrs = np.array([2 , 3 , 4 , 5 , 6])
marks = np.array([6 , 9 , 12 , 15 , 20])
```

```python
n = len(study_hrs)
m = (n*np.sum(study_hrs * marks) - np.sum(study_hrs)* np.sum(marks)) /  (n*np.sum(study_hrs**2) - (np.sum(study_hrs))**2)
c = (np.sum(marks)-m*np.sum(study_hrs))/n
print(m)
print(c)
```

```python
Y = m*study_hrs + c
print(Y)
```

```python
# Mean Squared Error
mse = np.mean((marks - Y)**2)

# Mean Absolute Error
mae = np.mean(np.abs(marks - Y))

# Root Mean Squared Error
rmse = np.sqrt(mse)
print("\nError Metrics:")
print("MSE:", mse)
print("MAE:", mae)
print("RMSE:", rmse)
```
