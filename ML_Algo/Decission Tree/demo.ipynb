```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
iris.feature_names
```

```python
iris.target_names
```

```python
print(iris.DESCR)
```

## Independent Features

```python
X = pd.DataFrame(iris['data'] , columns=iris.feature_names)
X.head()
```

## Dependent Feature

```python
y = pd.DataFrame(iris.target , columns=["target"])
y.head()
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(
    X , y , test_size=0.3 , random_state=42
)
```

## Applying Decission Tree Classifer

```python
from sklearn.tree import DecisionTreeClassifier
tree_classifier = DecisionTreeClassifier()
```

```python
tree_classifier.fit(X_train , y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(25 ,15))
tree.plot_tree(tree_classifier , filled=True)
```

```python
y_pred = tree_classifier.predict(X_test)
```

```python
from sklearn.metrics import classification_report , confusion_matrix , accuracy_score
```

```python
print(classification_report(y_test , y_pred))
```

```python
accuracy_score(y_test , y_pred)
```

## Decission Tree Classifier with Pre-Pruning and Hyper Parameter Tunning

```python
from sklearn.model_selection import RandomizedSearchCV
params = {
    "criterion": ['gini', 'entropy', 'log_loss'] , 
    "splitter" : ['best', 'random'] ,
    "max_depth" : [1 , 2 ,3 ,4 ,5] , 
    "max_features" : ['auto', 'sqrt', 'log2'] 
}
```

```python
random = RandomizedSearchCV(tree_classifier , param_distributions=params , cv = 5 , refit=True)
```

```python
random.fit(X_train , y_train)
```

```python
y_predict = random.predict(X_test)
```

```python
accuracy_score(y_test , y_predict)
```

```python
print(classification_report(y_test , y_predict))
```

```python
print(random.best_params_)
```
