## Train Test Split

## HyperParameter Tunning

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(X_train , y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score , classification_report , confusion_matrix
print(accuracy_score(y_test , y_pred))
print('confusion_matrix')
print(confusion_matrix(y_test , y_pred))
print('classification_report')
print(classification_report(y_test , y_pred))
```

### class_weight is used in LogisticRegression to give importance to classes (useful for imbalanced data).
### The code creates all combinations of weights for class 0 and class 1.
### Values [1, 10, 50, 100] are tried for both classes.
### Total combinations = 16 (4 × 4).
### Purpose: test different weights and find the best-performing one.

```python
# it gives the importamce and play with this numbers
class_weight = [{0:w , 1:y} for w in [1 , 10 , 50 , 100] for y in [1 , 10 , 50 , 100]]
class_weight
```

### Observation
The code generates 16 different class weight combinations to evaluate model performance under different importance levels of classes.

When both weights are equal ({0:1, 1:1}), the model behaves like a normal classifier without bias.

Increasing the weight of a class (e.g., {0:1, 1:100}) makes the model focus more on that class, improving its recall but possibly reducing overall accuracy.

If the dataset is imbalanced, higher weights for the minority class help in reducing bias toward the majority class.

Extremely large weights (like 100) may lead to overcompensation, causing misclassification of the other class.

Different weight combinations result in different precision-recall trade-offs, so the best choice depends on the problem goal (e.g., recall vs accuracy).

```python
penalty = ['l1' , 'l2' , 'elasticnet']
c_values = [100 , 10 , 1.0 , 0.1 , 0.01] 
solver = ['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga'] 
class_weight = [{0:w , 1:y} for w in [1 , 10 , 50 , 100] for y in [1 , 10 , 50 , 100]]
```

```python
params = dict(penalty=penalty , C=c_values , solver=solver , class_weight=class_weight)
```

```python
params
```

## Apply GridSearch CV and StratifiedKFold

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv = StratifiedKFold()
grid = GridSearchCV(estimator=model , param_grid=params , scoring='accuracy' , cv=cv)
```

```python
grid.fit(X_train , y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score , classification_report , confusion_matrix
print(accuracy_score(y_test , y_pred))
print('confusion_matrix')
print(confusion_matrix(y_test , y_pred))
print('classification_report')
print(classification_report(y_test , y_pred))
```
