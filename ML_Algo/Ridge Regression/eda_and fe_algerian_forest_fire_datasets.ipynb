## Algerian Forest Fires Dataset
Dataset Information
Additional Information

The dataset includes 244 instances that regroup a data of two regions of Algeria,namely the Bejaia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northwest of Algeria.

122 instances for each region. 

The period from June 2012 to September 2012. 
The dataset includes 11 attribues and 1 output attribue (class)
The 244 instances have been classified into â€˜fireâ€™ (138 classes) and â€˜not fireâ€™ (106 classes) classes.

### Additional Variable Information

1. Date : (DD/MM/YYYY) Day, month ('june' to 'september'), year (2012)
Weather data observations 
2. Temp : temperature noon (temperature max)  in Celsius degrees: 22 to 42
3. RH : Relative Humidity in %: 21 to 90 
4. Ws :Wind speed in km/h: 6 to 29 
5. Rain: total day in mm: 0 to 16.8
FWI Components  
6. Fine Fuel Moisture Code (FFMC) index from the FWI system: 28.6 to 92.5 
7. Duff Moisture Code (DMC) index from the FWI system: 1.1 to 65.9 
8. Drought Code (DC) index from the FWI system:  7 to 220.4
9. Initial Spread Index (ISI) index from the FWI system: 0 to 18.5 
10. Buildup Index (BUI) index from the FWI system: 1.1 to 68
11. Fire Weather Index (FWI) Index: 0 to 31.1
12. Classes: two classes, namely   â€œFireâ€ and â€œnot Fireâ€

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
dataset = pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv' , header=1)
dataset.head()
```

### Data Cleaning

```python
# mising Values
dataset.isnull().sum()
```

```python
# we do also like this to check the row of missing values
dataset[dataset.isnull().any(axis=1)]
```

The dataset is converted into two sets based on Regions from 122th index, we can make an new column. 

A new Column Based on the Region.

1 : Bejaia Region Dataset

2 : Sedi-Bel Abbes Region Dataset

Add new Column with Region

```python
dataset.loc[:122 , 'Region'] = 0
dataset.loc[122 : , 'Region'] = 1
df = dataset
```

```python
df.info()
```

Change the float value of Region Column to Int for memory management

```python
df['Region'] = df['Region'].astype('int')
```

```python
df.info()
```

```python
df.head()
```

### Checking null and missing Value

```python
df.isnull().sum()
```

### Removing null value

```python
df = df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

Checking the 122 index issue is solved or not

```python
df.iloc[[122]]
```

Dropping the 122th rows

```python
df = df.drop(122).reset_index(drop=True)
df.iloc[[122]]
```

```python
df.columns
```

```python
# Fix Space in Columns
df.columns = df.columns.str.strip()
```

```python
df.info()
```

```python
df.columns
```

Change the required Column as integer Data types

```python
df[['day' , 'month' , 'year' , 'Temperature' , 'RH' , 'Ws']] = df[['day' , 'month' , 'year' , 'Temperature' , 'RH' , 'Ws']].astype('int')
```

```python
df.info()
```

Changing the required Column as Float Data type

```python
objects = [feature for feature in df.columns if df[feature].dtypes=='O']
```

```python
for i in objects :
    if i != 'Classes' :
        df[i] = df[i].astype('float')
```

```python
# df[['Rain' , 'FFMC' , 'DMC' ,'DC' ,'ISI' ,'BUI' ,'FWI']] = df[['Rain' , 'FFMC' , 'DMC' ,'DC' ,'ISI' ,'BUI' ,'FWI']] .astype('float')
```
