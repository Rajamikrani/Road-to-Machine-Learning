```python
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
```

```python
x = np.arange(0 , 10)
y = np.arange(11 , 21)
```

```python
plt.scatter(x , y , c="r")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Graph in 2D")
plt.savefig("graph.png")
```

```python
y = x*x
```

```python
plt.plot(x , y , "r" , linestyle="dashed" , linewidth=2 , markersize=12)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Square of Number")
```

```python
plt.subplot(2 , 2 , 1)
plt.plot(x , y , "r")

plt.subplot(2 , 2 , 2)
plt.plot(x ,y , "r--")


plt.subplot(2 , 2 , 3)
plt.plot(x , y , "g--")

plt.subplot(2 , 2 , 4)
plt.plot(x ,y , "bo")
```

```python
x = np.arange(0 , 4 * np.pi , 0.1)
y = np.sin(x)
plt.plot(x , y , "r")
plt.title("Sine wave form")
```

```python
plt.subplot(2 , 1 , 1)
plt.plot(np.cos(x))
plt.title("Cosine")
plt.subplot(2 , 1 , 2)
plt.plot(np.sin(x))
plt.title("Sine")

```

## Bar Plot

```python
x = [2 , 6  , 10]
y = [3 , 6  , 16]
x2 = [3 , 4 , 7]
x1 = [4 , 8 , 14]
plt.bar(x , y)
plt.bar(x1 , x2 , color="r")
```

```python

data = [np.random.normal(0, std, 100) for std in range(1, 4)]

# rectangular box plot
plt.boxplot(data,vert=True,patch_artist=False);  
```

### Pie Chart

```python
labels = ["Python" , "Java" , "C++" ,"Ruby"]
sizes = [200 , 315 , 154 , 220]
colors = {"red" , "orange" , "yellow" , "blue"}
explode = [0.4 , 0 , 0 , 0]
plt.pie(sizes , explode=explode , labels=labels , colors=colors ,
        autopct='%1.1f%%', shadow=False)
plt.axis("equal")
```
