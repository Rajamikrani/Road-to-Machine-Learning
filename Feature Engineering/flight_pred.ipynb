```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_excel('Flight_price_prediction.xlsx')
df.head()
```

### Spliting Date_of_Journey Column into date , month and year column

```python
df['Date'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split("/").str[2]
```

```python
df.info()
```

### Converting Date,Month,Year column from object type to int type

```python
df['Date'] = df['Date'].astype('int')
df['Month'] = df['Month'].astype('int')
df['Year'] = df['Year'].astype('int')
```

```python
df.info()
```

```python
df.head()
```

### Droping the Date_of_Journey Columns

```python
df.drop('Date_of_Journey' , axis=1 , inplace = True)
```

```python
df.head(1)
```

### Spliting Arrival Time column to Arrival hours and Arrival Minutes

```python
df['Arrival_Time'] = df['Arrival_Time'].apply(lambda x:x.split(" ")[0])
```

```python
df['Arrival_Hrs'] = df['Arrival_Time'].str.split(':').str[0]
df['Arrival_Min'] = df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df.info()
```

### Converting Arrival_Hrs and arrival_Min column from object type to int type

```python
df['Arrival_Hrs'] = df['Arrival_Hrs'].astype('int')
df['Arrival_Min'] = df['Arrival_Min'].astype('int')
```

### Droping Arrival_Time Column

```python
df.drop("Arrival_Time" , axis=1 , inplace=True)
```
