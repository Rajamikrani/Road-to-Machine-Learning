```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("tips")
df.head()
```

## Coorelation with HeatMap
A correlation heatmap uses colored cells, typically in a monochromatic scale, to show a 2D correlation matrix (table) between two discrete dimensions or event types. It is very important in Feature Selection

```python
corr = df.select_dtypes(include="number").corr()
```

```python
corr
```

```python
sns.heatmap(corr , annot = True , cmap="coolwarm")
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

## JoinPlot
A Join Plot is allows to study the relationship between two numerical variables.
The central chart display the correlation it is usually a scatterplot , a hexbinplot , a 2-D histograms or a 2-D Density Plot.
Example :
sns.jointplot(x, y, data=df)
x → the column name from the DataFrame for the horizontal axis (X-axis)
y → the column name from the DataFrame for the vertical axis (Y-axis)

```python
sns.jointplot(x="tip" , y = "total_bill" , data=df , kind="hex")
```

```python
sns.jointplot(x="tip" , y = "total_bill" , data=df , kind="reg")
```

## PairPlot
A Pairplot is also known as Scattered Plot in which one variable in the same data row is matched with another variables value, like this pairplot are just elaborations on this. Showing all variable paired with all the other variables.

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue="sex")
```

```python
df['smoker'].value_counts()
```

```python
sns.pairplot(df , hue="smoker")
```

## Dist Plot
Dishplot helps to check the distribution of the column features

```python
sns.displot(df["tip"])
```

```python
sns.distplot(df['tip'] , kde=False , bins=10)
```

```python
sns.histplot(df['tip'])
```

## Box Plot
A box plot (also called box-and-whisker plot) in seaborn is used to summarize a numerical distribution using statistical quartiles.

```python
# Compare total bill distribution between male and female customers
sns.boxplot(data=df , x = "sex" , y= "total_bill")
```

```python
sns.boxplot(data=df , x='day' , y='total_bill' , palette='rainbow')
```

```python
sns.boxplot(data=df , x='day' , y='total_bill' , palette='rainbow' , hue='sex')
```

## Violin Plot
## ──────────────────────────────────────────────
#### ombines a box plot with a KDE (density curve).
#### The wider the violin, the more data points exist at that value range.
#### Useful for comparing distributions across categories with more detail than a box plot.
#### Compare total bill distribution across days, split by gender
#### The shape of each violin reveals where bills are most concentrated

#### Violion Plot help us to see both the distribution of data in terms of Kernel density estimation (KDE) and the box 

```python
sns.violinplot(data=df , x = 'total_bill' , y = 'day' , palette='rainbow' , hue="sex" )
```
