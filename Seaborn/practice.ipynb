```python
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
df.head()
```

## Correlation with Hashmap

```python
corr = df.select_dtypes(include="number" ).corr()
corr
```

```python
sns.heatmap(corr , annot=True , cmap="coolwarm")
```

## Join Plot

```python
sns.jointplot(data=df , x='sepal_length' , y = 'petal_length', kind='reg')
```

```python
sns.jointplot(data=df , x='sepal_length' , y = 'petal_length', kind='hex')
```

## PairPlot

```python
sns.pairplot(df)
```

```python
df['species'].value_counts()
```

```python
sns.pairplot(data=df , hue="species")
```

```python
sns.displot(df["sepal_width"])
```

```python
sns.distplot(df['sepal_width'] , kde=False , bins=10)
```

```python
sns.histplot(df['sepal_width'])
```

## Categorial Plot

```python
sns.countplot(data=df , x="species")
```

```python
sns.countplot(data=df , y = "species")
```

```python
sns.countplot(data=df , x="sepal_width")
```

## Bar Plot

```python
sns.barplot(data=df , x='sepal_length' , y='species')
```
